#### Ejercicio 01. ¿Cuál es la casa más cara en todo el dataset? (★☆☆)

Imprime la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle del Prado, Nº20 es la más cara y su precio es de 5000000 USD`

```python
# TODO
# Listo el nombre de las columnas del dataframe
df.columns.tolist()

#  Busco el indice de la fila con el precio más alto 
id_max = df["price"].idxmax()

#  con .loc  y  con el id maximo busco su direccion  y precio de la fila  
address = df.loc[id_max, "address"]
price = df.loc[id_max, "price"]

#  lo imprimo 
print(f"La casa con direccion en {address} es la mas cara y su precio es de {price} USD")
```

#### Ejercicio 02. ¿Cuál es la casa más barata del dataset? (★☆☆)

Este ejercicio es similar al ejercicio anterior, solo que ahora buscamos la casa con el precio más bajo. Recuerda imprimir la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle Alcalá, Nº58 es la más barata y su precio es de 12000 USD`

```python
# Primero filtro que el precio debe ser mayor a 0 
data_mayor_0 = df[df["price"] > 0]

# Busco el indice de la fila   con el precio minimo
id_min = data_mayor_0["price"].idxmin()

# Con el .loc y el indice  busco la direccion y el precio de la fila 
address_min = data_mayor_0.loc[id_min, "address"]
price_min = data_mayor_0.loc[id_min, "price"]

# Imprimo el resultado
print(f"La casa con direccion en {address_min} es la mas barata y su precio es de {price_min} USD")
```

#### Ejercicio 03. ¿Cuál es la casa más grande y la más pequeña del dataset? (★☆☆)

Imprime la dirección y el área de las casas seleccionadas. Por ejemplo:

`La casa más grande está ubicada en Calle Gran Vía, Nº38 y su superficie es de 5000 metros`

`La casa más pequeña está ubicada en Calle Mayor, Nº12 y su superficie es de 200 metros`

Este ejercicio es similar al anterior, pero buscamos las casas más grandes y pequeñas en base a su superficie.

```python
# Primero filtro que los metros sean mayores a 0 
data_surface_0 = df[df["surface"] > 0]

#  Busco el indice con los metros mas bajos y altos 
id_min = data_surface_0["surface"].idxmin()
id_max = data_surface_0["surface"].idxmax()

# Y con .loc y el indice localizo la calle y la superfice mas mayor y mas baja
address_min = data_surface_0.loc[id_min, "address"]
surface_min = data_surface_0.loc[id_min, "surface"]

address_max = data_surface_0.loc[id_max, "address"]
surface_max = data_surface_0.loc[id_max, "surface"]

# Imprimo el resultado
print(f"La casa más grande está ubicada en {address_max} y su superficie es de {surface_max} metros")
print(f"La casa más pequeña está ubicada en {address_min} y su superficie es de {surface_min} metros")
```

#### Ejercicio 08. Trazar el histograma de los precios para la población de "Arroyomolinos (Madrid)" y explica qué observas (★★☆)

Imprime el histograma de los precios y escribe en la celda del Markdown un breve análisis del trazado.


```python
import matplotlib.pyplot as plt

# Primero filtro por la columna level 5 con nombre de la fila Arroyomolinos y selecciono la columna precios de esa fila 
arroyomolinos = df[df["level5"] == "Arroyomolinos (Madrid)"]
precios = arroyomolinos["price"]

# Dibuja el histograma con 10  de color rosa y negro
plt.hist(precios, color="pink", edgecolor="black")  
# Pongo titulo 
plt.title("Histograma de precios - Arroyomolinos (Madrid)")
# Nombre del eje X
plt.xlabel("Precio")
# Nombre del eje Y
plt.ylabel("Cantidad de casas")
# Muestro el grafico 
plt.show()
```
